## 1. Understand LLMs

### 1.1 What is an LLM?
Simple terms, an LLM is a system trained to predict the next word given a previous input or text. i.e. Input -> BikeEase is the best way to -> LLM predicts -> commute around the city quickly and affordably. This is a learned pattern matching for the given input on what is most likely to follow.

### 1.2 Why an LLM?
Marketing is persuasion, tone, targeting audience and creativity. LLMs are amazing at copying tone, creating variations and personalizing. This makes them exclusively tailored to excel at this type of problem.
